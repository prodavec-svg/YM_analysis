# Конвертер Parquet -> CSV

```
import pandas as pd
from pathlib import Path

MARTS_DIR = Path("data/marts")
CSV_DIR = Path("data/saved_csv")
CSV_DIR.mkdir(parents=True, exist_ok=True)
```

## Настройки

```
# Укажите название витрины без расширения (например, "mart_user_segments")
# или установите значение "all", чтобы конвертировать все витрины сразу.
target_mart = "all"
```

## Запуск конвертации

```
def convert_parquet_to_csv(parquet_path: Path):
    if not parquet_path.exists():
        print(f"Файл не найден: {parquet_path}")
        return
    
    df = pd.read_parquet(parquet_path)
    csv_path = CSV_DIR / f"{parquet_path.stem}.csv"
    df.to_csv(csv_path, index=False)
    print(f"Конвертирован: {parquet_path.name} -> {csv_path.name}")

if target_mart.lower() == "all":
    parquet_files = list(MARTS_DIR.glob("*.parquet"))
    if not parquet_files:
        print(f"Нет .parquet файлов в {MARTS_DIR}")
    else:
        for p in parquet_files:
            convert_parquet_to_csv(p)
else:
    # Конвертируем только указанную витрину
    specific_path = MARTS_DIR / f"{target_mart}.parquet"
    convert_parquet_to_csv(specific_path)

print("Готово.")
```
