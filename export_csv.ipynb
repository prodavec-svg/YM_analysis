# DataLens CSV Exporter
Конвертирует собранные Parquet-витрины в CSV-формат для загрузки в Yandex DataLens.

```python
import polars as pl
from pathlib import Path

PROJECT_DIR = Path.cwd().resolve()
MARTS_DIR = PROJECT_DIR / "data" / "marts"
CSV_DIR = PROJECT_DIR / "data" / "saved_csv"

CSV_DIR.mkdir(parents=True, exist_ok=True)

print(f"Найдено витрин: {len(list(MARTS_DIR.glob(*.parquet)))}")
```

```python
for parquet_path in MARTS_DIR.glob("*.parquet"):
    csv_path = CSV_DIR / parquet_path.name.replace(".parquet", ".csv")
    print(f"Конвертация {parquet_path.name} в CSV...")
    # Используем streaming для избежания проблем с памятью на больших витринах
    pl.scan_parquet(parquet_path).sink_csv(csv_path)

print("✅ Все витрины успешно экспортированы в data/saved_csv/")
```
